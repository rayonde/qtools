# S15 TDC1：获取 histogram 和指定窗口符合计数

这个示例使用 `TDC` 上下文管理器，需要连接 S-Fifteen TDC1。
结果中的时间轴使用 ps。

```python
import matplotlib.pyplot as plt

from qtools.tdc import TDC

BACKEND = "exp_s15_tdc1"
BIN_WIDTH = 2_000
```

## 1. 获取 histogram


```python
with TDC(backend=BACKEND, bin_width=BIN_WIDTH) as tdc:
    histogram_result = tdc.measure_g2(
        duration=1.0,
        samples=1,
        ch_start=1,
        ch_stop=2,
        ch_stop_delay=BIN_WIDTH * 5,
        unit="ps",
        bins=500,
    )

histogram = histogram_result.histogram[0]
edges_ps = histogram_result.bin_edges[0]
centers_ps = (edges_ps[:-1] + edges_ps[1:]) / 2

plt.step(centers_ps, histogram, where="mid")
plt.xlabel("delay (ps)")
plt.ylabel("counts")
plt.show()
```

## 2. 获取指定 window 内的符合计数


```python
with TDC(backend=BACKEND, bin_width=BIN_WIDTH) as tdc:
    window_result = tdc.measure_coincidence(
        duration=1.0,
        samples=1,
        ch_start=1,
        ch_stop=2,
        unit="ps",
        window_start=0,
        window_stop=10_000,
    )

print("window:", window_result.window_start[0], "to", window_result.window_stop[0], "ps")
print("raw coincidences:", window_result.raw_counts[0])
```
